```python
import numpy as np
import pandas as pd
```

Merging 2 data frames

```python
# First DataFrame - Students
students = {
    "emp_id":[1,2,3,10,7],
    "Name": ["Manisha", "Sayali", "Sanika", "Sai", "Shreya"],
    "Marks": [85, 92, 89, 76, 88],
    "Age": [21, 22, 20, 19, 23]
}
pd.DataFrame(students)
```

```python
courses = {
    "emp_id":[1,2,8,9,4],
    "Course": ["Manisha", "Sayali", "Sanika", "Sai", "Shreya"],
    "Duration_Months": [3, 4, 6, 8, 5],
    "Fees": [15000, 18000, 25000, 30000, 40000]
}
pd.DataFrame(courses)
```

```python
df_students = pd.DataFrame(students)
```

```python
df_courses = pd.DataFrame(courses)
```

```python
merged_df = pd.merge(df_students, df_courses, on="emp_id",how="inner")
merged_df
```

```python
merged_df = pd.merge(df_students, df_courses, on="emp_id",how="outer")
merged_df
```

```python
merged_df = pd.merge(df_students, df_courses, on="emp_id",how="left")
merged_df
```

```python
merged_df = pd.merge(df_students, df_courses, on="emp_id",how="right")
merged_df
```

```python
Concatanation of 2 DataFrames
```

```python
students_1 = {
    "emp_id": [1, 2, 3],
    "Name": ["Manisha", "Sayali", "Sanika"],
    "Marks": [85, 92, 89]
}
df1=pd.DataFrame(students_1)
```

```python
df1
```

```python
students_2 = {
    "emp_id": [4, 5, 6],
    "Name": ["Sai", "Shreya", "Shubhangi"],
    "Marks": [76, 88, 91]
}
df2=pd.DataFrame(students_2)
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1) #On the basis of rows
```

```python
pd.concat([df1,df2],axis=0) #On the basis of column
```
