```python
import numpy as np 
import pandas as pd
```

```python
GroupBY #groupby() is a pandas function used to divide data into groups based on one or more columns.
```

```python
data = {
    "Name": ["Manisha", "Sayali", "Sanika", "Shreya", "Sai"],

    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai"],

    "Course": ["BTech", "BCA", "BTech", "BSc", "BCA"],

    "Marks": [85, 78, 92, 74, 81],

    "Grade": ["A", "B", "A+", "B", "A"]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
#GroupBY Grades and calculate the sum of marks
cat=df.groupby("Grade")#Group by grades
cat
for i,v in cat:
    print(i)
    print(v)
```

```python
cat=df.groupby("Grade") ['Marks'].sum() #sum of marks
cat
```

```python
#Group by multiple columns
#group by course and grade
cat=df.groupby(['Course','Grade']) ['Marks'].sum()
cat
```

```python
Aggregation #Aggregation means performing calculations on grouped data such as:sum(),mean(),max(),min(),count()
```

```python
df['Marks'].mean()
```
