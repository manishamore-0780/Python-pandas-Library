```python
import numpy as np
import pandas as pd
```

```python
Creating a DataFrames
```

```python
data = {
    'Name':['Manisha','Sayali','Sanika','Sai'],
    'Age':[18,19,19,20],
    'Department':['CS','IT','ME','ENTC'],
    'Marks':[98,86,97,78]
}
df1 = pd.DataFrame(data)
```

```python
data_list = [
    ['Manisha',18,'CS',98],
    ['Sayali',19,'IT',86],
    ['Sanika',19,'ME',97],
    ['Sai',20,'ENTC',88]
]
df2=pd.DataFrame(data_list)
columns=['Name','Age','Department','Marks']
df2=pd.DataFrame(data_list,columns=columns)
```
