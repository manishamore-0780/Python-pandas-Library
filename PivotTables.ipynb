```python
import numpy as np
import pandas as pd
```

```python
Pivot Table #A Pivot Table is used to summarize, organize, and analyze data in a table format.
#Pivot tables make large data easy to understand.
```

```python

data = {
    "Name": ["Manisha", "Sayali", "Sanika", "Shreya", "Sai"],
    "City": ["Pune", "Mumbai", "Pune", "Nashik", "Mumbai"],
    "Marks": [8,7,9,7,1],
    "Grade": ["A", "B", "A+", "B", "A"],
    "Percentage":[89,90,91,77,87]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
pivot1=pd.pivot_table(df,values='Marks',index='Grade',columns='Name')
pivot1
```

```python
pd.pivot_table(df,values='Marks',index='Grade',columns='Name',aggfunc='median')
```

```python
pivot2=pd.pivot_table(df,values=['Marks','Percentage'],index='Grade',columns='Name',aggfunc='median')
pivot2
```

```python
Cross Tabs #crosstab() is used to calculate the frequency (count) between two or more columns.
#It creates a table showing how many times categories occur together.
#It is similar to a Pivot Table, but mainly used for counting data.
```

```python
pd.crosstab(df['City'],['Name'])
```
